```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Set gamma value
gamma = 0.4  # <1 to brighten an underexposed image

# Read the original image
img_original = cv2.imread("room-light.jpg")

# Create a lookup table for gamma correction
lookUpTable = np.empty((1, 256), np.uint8)
for i in range(256):
    lookUpTable[0, i] = np.clip(pow(i / 255.0, gamma) * 255.0, 0, 255)

# Apply gamma correction using the lookup table
out = cv2.LUT(img_original, lookUpTable)

# Display original and corrected images side by side
plt.figure(figsize=[15, 5])
plt.subplot(121)
plt.imshow(img_original[..., ::-1])  # Convert BGR → RGB
plt.title("Original")
plt.subplot(122)
plt.imshow(out[..., ::-1])
plt.title("Gamma Corrected")
plt.show()
```

```python
gamma = 0.4  

img_original = cv2.imread("room-light.jpg")

output_linear = cv2.convertScaleAbs(img_original, alpha=1.3, beta=40)

# Prepare lookup table for gamma correction
lookUpTable = np.empty((1, 256), np.uint8)
for i in range(256):
    lookUpTable[0, i] = np.clip(pow(i / 255.0, gamma) * 255.0, 0, 255)

# Apply gamma correction
output_gammacorrected = cv2.LUT(img_original, lookUpTable)

# Display original, linear, and gamma-corrected images side by side
plt.figure(figsize=[17,6])

plt.subplot(231)
plt.imshow(img_original[..., ::-1])
plt.title("Original")
plt.subplot(232)
plt.imshow(output_linear[..., ::-1])
plt.title("Linear")
plt.subplot(233)
plt.imshow(output_gammacorrected[..., ::-1])
plt.title("Gamma")

plt.show()
```
