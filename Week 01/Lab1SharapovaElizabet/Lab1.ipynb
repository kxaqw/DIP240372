Step 2: Reading and Converting to Grayscale

Method A: Converting without cvtColor (Direct Read)

```python
# Read and convert the image to grayscale directly
gray_direct = cv2.imread('fb.jpg', cv2.IMREAD_GRAYSCALE)
gray_direct1 = cv2.imread('landscape1.jpg', cv2.IMREAD_GRAYSCALE)
gray_direct2 = cv2.imread('landscape2.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
# Module to display the image
cv2.imshow('Grayscale - Direct Read', gray_direct)
cv2.imshow('Grayscale - Direct Read1', gray_direct1)
cv2.imshow('Grayscale - Direct Read2', gray_direct2)
# Module to hold the displaying window
cv2.waitKey(0)
```

```python
# Save the result
cv2.imwrite('gray_direct.jpg', gray_direct)
cv2.imwrite('gray_direct1.jpg', gray_direct1)
cv2.imwrite('gray_direct2.jpg', gray_direct2)
```

Method B: Converting with cvtColor module

```python
# Convert from BGR to Grayscale using cvtColor
gray_cvt = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_cvt1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray_cvt2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
```

```python
# Module to display the image
cv2.imshow('Grayscale - Converted', gray_cvt)
cv2.imshow('Grayscale - Converted1', gray_cvt1)
cv2.imshow('Grayscale - Converted2', gray_cvt2)
# Module to hold the displaying window
cv2.waitKey(0)
```

```python
# Save the result
cv2.imwrite('gray_cvt.jpg', gray_cvt)
cv2.imwrite('gray_cvt1.jpg', gray_cvt1)
cv2.imwrite('gray_cvt2.jpg', gray_cvt2)
```

Task 3: Thresholding Experiments
1 & 2. Experiment with different threshold values and Save results

Step 3: Thresholding to Create a Binary Image

```python
# --- Experiment 1: Low Threshold (100) ---
_, binary_100 = cv2.threshold(gray_cvt, 100, 255, cv2.THRESH_BINARY)
_, binary1_100 = cv2.threshold(gray_cvt1, 100, 255, cv2.THRESH_BINARY)
_, binary2_100 = cv2.threshold(gray_cvt2, 100, 255, cv2.THRESH_BINARY)
cv2.imwrite('binary_100.jpg', binary_100) # Save image
cv2.imwrite('binary_100.1.jpg', binary1_100)
cv2.imwrite('binary_100.2.jpg', binary2_100)
```

```python
# --- Experiment 2: Mid Threshold (128) ---
_, binary_128 = cv2.threshold(gray_cvt, 128, 255, cv2.THRESH_BINARY)
_, binary1_128 = cv2.threshold(gray_cvt1, 128, 255, cv2.THRESH_BINARY)
_, binary2_128 = cv2.threshold(gray_cvt2, 128, 255, cv2.THRESH_BINARY)
cv2.imwrite('binary_128.jpg', binary_128) # Save image
cv2.imwrite('binary_128.1.jpg', binary1_128)
cv2.imwrite('binary_128.2.jpg', binary2_128)
```

```python
# --- Experiment 3: High Threshold (180) ---
_, binary_180 = cv2.threshold(gray_cvt, 180, 255, cv2.THRESH_BINARY)
_, binary1_180 = cv2.threshold(gray_cvt1, 180, 255, cv2.THRESH_BINARY)
_, binary2_180 = cv2.threshold(gray_cvt2, 180, 255, cv2.THRESH_BINARY)
cv2.imwrite('binary_180.jpg', binary_180) # Save image
cv2.imwrite('binary_180.1.jpg', binary1_180)
cv2.imwrite('binary_180.2.jpg', binary2_180)
```

```python
# Saving the binary images
cv2.imwrite('binary_100.jpg', binary_100)
cv2.imwrite('binary_100.1.jpg', binary1_100)
cv2.imwrite('binary_100.2.jpg', binary2_100)
cv2.imwrite('binary_128.jpg', binary_128)
cv2.imwrite('binary_128.1.jpg', binary1_128)
cv2.imwrite('binary_128.2.jpg', binary2_128)
cv2.imwrite('binary_180.jpg', binary_180)
cv2.imwrite('binary_180.1.jpg', binary1_180)
cv2.imwrite('binary_180.2.jpg', binary2_180)
```
