```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

Part 1: Arithmetic Operations on Images
Step 1: Subtracting an Intensity from an Image
Task 1: 
1. Experiment with different intensity values (e.g., 30, 100) and observe how the image
changes.

```python
image = cv2.imread('dog_img.jpg')
image_RGB = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(image_RGB)
axes[0].set_title('Original')
axes[0].axis('off')

subtracted_30 = cv2.subtract(image_RGB, 30)
axes[1].imshow(subtracted_30)
axes[1].set_title('Sustracted 30')
axes[1].axis('off')

subtracted_100 = cv2.subtract(image_RGB, 100)
axes[2].imshow(subtracted_100)
axes[2].set_title('Sustracted 100')
axes[2].axis('off')

plt.tight_layout()
plt.savefig('task1_subtraction.png', dpi=150)
plt.show()
```

2. Explain why the image becomes darker as you subtract more.

Subtracting reduces pixel intensity values (0–255), moving them toward 0, which makes the image darker. Larger subtraction results in darker pixels; values below 0 are clipped to black.

3. The provided examples are for subtracting an intensity from all three RGB slices. Write a
source code that reduce the intensities of the Red channel only.

```python
def subtract_red_channel(image, value):
    image_copy = image.copy()
    image_copy[:, :, 0] = cv2.subtract(image_copy[:, :, 0], value)
    return image_copy

red_subtracted = subtract_red_channel(image_RGB, 100)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(image_RGB)
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(red_subtracted)
axes[1].set_title('Red Channel -100')
axes[1].axis('off')

plt.tight_layout()
plt.savefig('task1_red_channel.png', dpi=150)
plt.show()
```

Step 2: Adding an Intensity to an Image

Task 2: 
1. Experiment with different intensity values (e.g., 30, 100) and observe how the image
changes.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(image_RGB)
axes[0].set_title('Original')
axes[0].axis('off')

added_30 = cv2.add(image_RGB, 30)
axes[1].imshow(added_30)
axes[1].set_title('Added 30')
axes[1].axis('off')

added_100 = cv2.add(image_RGB, 100)
axes[2].imshow(added_100)
axes[2].set_title('Added 100')
axes[2].axis('off')

plt.tight_layout()
plt.savefig('task2_addition.png', dpi=150)
plt.show()
```

```python
2. Explain why the image becomes brighter as you add more.
```

Adding increases pixel intensity values toward 255, making the image brighter. Larger additions increase brightness; values above 255 are clipped to white.

3. The provided examples are for adding an intensity from all three RGB slices. Write a
source code that increases the intensities of the Red channel only.

```python
def add_red_channel(image, value):
    image_copy = image.copy()
    image_copy[:, :, 0] = cv2.add(image_copy[:, :, 0], value)
    return image_copy

red_added = add_red_channel(image_RGB, 100)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(image_RGB)
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(red_added)
axes[1].set_title('Red chanell +100')
axes[1].axis('off')

plt.tight_layout()
plt.savefig('task2_red_channel.png', dpi=150)
plt.show()
```

Step 3: Multiplying an Image by a Constant

```python
Task 3:
1. Experiment with different multiplication factors (e.g., 0.5, 2.0) and observe how the
image contrast changes.
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(image_RGB)
axes[0].set_title('Original (×1.0)')
axes[0].axis('off')

multiplied_05 = cv2.multiply(image_RGB.astype(np.float32), 0.5)
multiplied_05 = np.uint8(np.clip(multiplied_05, 0, 255))
axes[1].imshow(multiplied_05)
axes[1].set_title('Multiplied 0.5')
axes[1].axis('off')

multiplied_20 = cv2.multiply(image_RGB.astype(np.float32), 2.0)
multiplied_20 = np.uint8(np.clip(multiplied_20, 0, 255))
axes[2].imshow(multiplied_20)
axes[2].set_title('Multiplied 2.0')
axes[2].axis('off')

plt.tight_layout()
plt.savefig('task3_multiplication.png', dpi=150)
plt.show()
```

2. Discuss the impact of multiplying the pixel values by a number greater than 1 and less
than 1.

Multiplier > 1 → increases brightness and contrast.
Multiplier < 1 → decreases brightness and contrast.

Step 4: Dividing an Image by a Constant

Task 4:
1. Experiment with different division factors (e.g., 3.0, 0.5).

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

axes[0].imshow(image_RGB)
axes[0].set_title('Original (÷1.0)')
axes[0].axis('off')

divided_05 = cv2.divide(image_RGB.astype(np.float32), 0.5)
divided_05 = np.uint8(np.clip(divided_05, 0, 255))
axes[1].imshow(divided_05)
axes[1].set_title('Divided 0.5')
axes[1].axis('off')

divided_20 = cv2.divide(image_RGB.astype(np.float32), 2.0)
divided_20 = np.uint8(np.clip(divided_20, 0, 255))
axes[2].imshow(divided_20)
axes[2].set_title('Divided 2.0')
axes[2].axis('off')

divided_30 = cv2.divide(image_RGB.astype(np.float32), 3.0)
divided_30 = np.uint8(np.clip(divided_30, 0, 255))
axes[3].imshow(divided_30)
axes[3].set_title('Divided 3.0')
axes[3].axis('off')

plt.tight_layout()
plt.savefig('task4_division.png', dpi=150)
plt.show()
```

2. Compare the effects of dividing by a number greater than 1 versus less than 1.

Dividing by a number > 1 → decreases brightness and contrast.
Dividing by a number < 1 → increases brightness and contrast.
